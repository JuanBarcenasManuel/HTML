```python
import pandas as pd
import requests
import sqlite3
import re
from bs4 import BeautifulSoup 
from datetime import datetime

url = "https://en.wikipedia.org/wiki/List_of_most-streamed_songs_on_Spotify"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

try:
    response = requests.get(url, headers=headers)
    response.raise_for_status() 
    
    html_text = response.text
    print("Descarga exitosa (código 200 OK).")
    
except requests.exceptions.RequestException as e:
    print(f"Error al descargar la página: {e}")
    exit()

try:
    dataframes = pd.read_html(html_text, flavor='lxml')
    
    df = dataframes[0].copy()
    print(f"Tabla cargada exitosamente. Filas iniciales: {len(df)}")
    
except ValueError as e:
    print(f"Error al leer la tabla HTML con pandas: {e}")
    exit()


df.columns = ['Rank', 'Song', 'Artist', 'Streams (Billions)', 'Release Date', 'Notes']


df = df[pd.to_numeric(df['Rank'], errors='coerce').notna()]

def clean_streams(stream_value):
    cleaned = re.sub(r'\[.*?\]', '', str(stream_value)).strip()
    return float(cleaned)

df['Streams'] = df['Streams (Billions)'].apply(clean_streams)
df.drop(columns=['Streams (Billions)', 'Notes'], inplace=True)

def extract_year(date_str):
    match = re.search(r'(\d{4})', str(date_str))
    return int(match.group(1)) if match else None

df['Release Year'] = df['Release Date'].apply(extract_year)
df.drop(columns=['Release Date'], inplace=True)

df['Rank'] = pd.to_numeric(df['Rank'], errors='coerce').astype('Int64')

df = df[['Rank', 'Song', 'Artist', 'Streams', 'Release Year']]

print("\n--- DataFrame Limpio (Primeras 5 Filas) ---")
print(df.head())

db_name = 'spotify_rankings.sqlite'
table_name = 'most_streamed_songs'
conn = sqlite3.connect(db_name)
cursor = conn.cursor()

print(f"\nAlmacenando datos en la base de datos '{db_name}'...")

df.to_sql(table_name, conn, if_exists='replace', index=False)

conn.commit()

result = cursor.execute(f"SELECT COUNT(*) FROM {table_name}").fetchone()
print(f"Datos almacenados con éxito. Total de filas en DB: {result[0]}")

conn.close()
```
